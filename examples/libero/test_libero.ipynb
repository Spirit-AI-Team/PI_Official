```python
import tensorflow_datasets as tfds
```

```python
RAW_DATASET_NAMES = [
    "libero_10_no_noops",
    "libero_goal_no_noops",
    "libero_object_no_noops",
    "libero_spatial_no_noops",
]  # For simplicity we will combine multiple Libero datasets into one training dataset

dataset = tfds.load(RAW_DATASET_NAMES[0], data_dir='/hy-tmp/likaiyu/resources/modified_libero_rlds', split = 'train')
```

```python
data = [epi for epi in dataset]
```

```python
len(data)
```

```python
epis = []
for d in data:
    steps = [step for step in d["steps"].as_numpy_iterator()]
    epis.append(steps)
```

```python
l = 0
l_max = 0
for cnt, steps in enumerate(epis):
    prompt = steps[0]['language_instruction']
    l += len(prompt.split())
    l_max = max(len(prompt.split()), l_max)
    if cnt % 50 == 0:
        print(prompt)
print(f'mean len is {l/len(epis)}')
print(f'max len is {l_max}')
```
